```python
import pandas as pd


data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [52000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [5, 8, 10, 3, 12],
    'Joining_Date': ['2020-05-15', '2017-07-19', '2013-06-01',
                     '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.head(5))
print(df.tail(5))
```

```python
print(df['Salary'].mean())
print(df['Bonus'].sum())
print(df['Age'].min())
print(df['Rating'].max())
```

```python
df = df.sort_values('Salary', ascending=False)
print(df)
```

```python
def performance(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

df['Performance'] = df['Rating'].apply(performance)
print(df)

print(df.isnull().sum())
        
```

```python
df = df.rename(columns={'Employee_ID': 'ID'})
print(df)
```

```python
print(df[df['Years_of_Experience'] > 5])
```

```python
df['Tax'] = df['Salary'] * 0.10

print("\nFinal DataFrame:")
print(df)
```

```python
df.to_csv('employees_modified.csv', index=False)

print("\nFile saved successfully!")
```
