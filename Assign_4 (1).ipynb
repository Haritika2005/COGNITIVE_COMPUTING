```python
#2

import pandas as pd

def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():

        # Combine question and keywords
        text = row["question"].lower() + " " + row["keywords"].lower()
        entry_words = set(text.split())

        # Find matching words
        matching_words = query_words.intersection(entry_words)

        # Calculate confidence
        if len(query_words) > 0:
            confidence = len(matching_words) / len(query_words)
        else:
            confidence = 0

        # Keep only matching entries
        if confidence > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })

    # Sort by confidence (highest first)
    results = sorted(
        results,
        key=lambda x: x["confidence"],
        reverse=True
    )

    return results


# Test the function
query = "how can i pay my fee"

matches = score_query(query, df)

for result in matches:
    print(result)

                      
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():

        text = row["question"].lower() + " " + row["keywords"].lower()
        entry_words = set(text.split())

        matching_words = query_words.intersection(entry_words)

        if len(query_words) > 0:
            confidence = len(matching_words) / len(query_words)
        else:
            confidence = 0

        if confidence > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })

    # Sort from highest confidence to lowest
    results = sorted(
        results,
        key=lambda x: x["confidence"],
        reverse=True
    )

    # Check highest score
    if results:
        highest_score = results[0]["confidence"]

        print("Highest confidence:", highest_score)

        # Print ALL entries having highest score
        for result in results:
            if result["confidence"] == highest_score:
                print(result)

    return results
    # Tie demonstration
print("TIE QUERY:")
score_query("fee", df)

# Non-tie demonstration
print("\nNON-TIE QUERY:")
score_query("reset password", df)
    
```
