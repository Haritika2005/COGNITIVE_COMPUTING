```python
import re
import nltk
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
text = """Leh Ladakh is a beautiful place in India.
It is famous for its high mountains, blue lakes, and peaceful surroundings.
Many tourists visit Ladakh to enjoy its natural beauty.
Pangong Lake and Nubra Valley are popular tourist attractions.
The weather is cold, but the scenery is amazing.
I would love to visit Leh Ladakh and explore its beautiful landscapes."""

print("Original text:",text)

clean_text=re.sub(r'[^\w\s]','',text.lower())
print("\n lower case text without pnctuation",clean_text)
words=word_tokenize(clean_text)
sentences=sent_tokenize(text)

print(clean_text.split())
print(words)

stop_words=set(stopwords.words('english'))
filtered_words=[word for word in words if word not in stop_words]
freq=FreqDist(filtered_words)
for word,count in freq.items():
    print(word,":",count)
```

```python

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Same paragraph from Q1
text = """Leh Ladakh is a beautiful place in India.
It is famous for its high mountains, blue lakes, and peaceful surroundings.
Many tourists visit Ladakh to enjoy its natural beauty.
Pangong Lake and Nubra Valley are popular tourist attractions.
The weather is cold, but the scenery is amazing.
I would love to visit Leh Ladakh and explore its beautiful landscapes."""

# 1. Convert to lowercase and extract alphabetic words
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print("Alphabetic Words:\n", words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]
print("\nAfter Stopword Removal:\n", filtered_words)

# 3. Perform stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("\nStemmed Words:\n", stemmed_words)

# 4. Perform lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_words
]
print("\nLemmatized Words:\n", lemmatized_words)

# 5. Compare stemming and lemmatization
print("\nComparison:")
for word, stem, lemma in zip(filtered_words, stemmed_words, lemmatized_words):
    print(word, "-> Stem:", stem, "| Lemma:", lemma)
```
