```python
#1a
import numpy as np
arr=np.array([1,2,3,4,5])
print("after adding 2=",arr+2)
print("after multiplying with 3=",arr*3)
print("after dividing with 2=",arr/2)
```

```python
#2
import numpy as np
arr=np.array([1,2,3,6,4,5])
print("original array",arr)
print("reversed array",arr[:: -1])
```

```python
#2ii
import numpy as np
x = np.array([1,2,3,4,5,1,2,1,1,1])
values,count=np.unique(x, return_counts=True)
most_freq=values[np.argmax(count)]
print(most_freq)
```
