```python
#3
#i. 
import random
random.seed(1024170015)

#ii. Generate a list of 100 random numbers between 100 and 900 (inclusive). Count and print all odd numbers in the list.
numbers=[random.randint(100,900) for i in range(100)]
print(numbers)
odd=[x for x in numbers if x%2!=0]
print(odd)
print(len(odd))

    

#iii. Count and print all even numbers in the list.
even=[x for x in numbers if x%2==0 ]
print(even)
print(len(even))
#     iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension. 
def is_prime(x):
    f=0
    for i in range(1,x+1):
        if x%i==0:
            f=f+1
    if f==2:
        return True
    else:
        return False

prime_l=[ x for x in numbers if is_prime(x)]
print(prime_l)
print(len(prime_l))
            
    
            

 # v. Print the number that occurs most frequently in your list, and how many times it occurs. 

most_frequent = numbers[0]
max_count = numbers.count(numbers[0])

for x in numbers:
    if numbers.count(x) > max_count:
        max_count = numbers.count(x)
        most_frequent = x

print("Most frequent number:", most_frequent)
print("Number of times it occurs:", max_count)
```
